```python
# reward_utils.py
# Utility functions to compute ROUGE-based reward for reinforcement learning

from rouge_score import rouge_scorer


def compute_rouge_scores(pred_summary, ref_summary):
    """
    Compute ROUGE-1, ROUGE-2, and ROUGE-L F1 scores.

    Args:
        pred_summary (str): The predicted/generated summary.
        ref_summary (str): The reference/ground truth summary.

    Returns:
        dict: A dictionary containing ROUGE-1, ROUGE-2, and ROUGE-L F1 scores.
    """
    scorer = rouge_scorer.RougeScorer(['rouge1', 'rouge2', 'rougeL'], use_stemmer=True)
    scores = scorer.score(ref_summary, pred_summary)
    return {
        'rouge1': scores['rouge1'].fmeasure,
        'rouge2': scores['rouge2'].fmeasure,
        'rougeL': scores['rougeL'].fmeasure
    }


def compute_reward(pred_summary, ref_summary, metric="rougeL"):
    """
    Compute scalar reward based on specified ROUGE metric (default is ROUGE-L).

    Args:
        pred_summary (str): Generated summary.
        ref_summary (str): Reference summary.
        metric (str): One of 'rouge1', 'rouge2', or 'rougeL'.

    Returns:
        float: Reward score (F1 value of selected ROUGE metric).
    """
    scores = compute_rouge_scores(pred_summary, ref_summary)
    return scores.get(metric, 0.0)
```
